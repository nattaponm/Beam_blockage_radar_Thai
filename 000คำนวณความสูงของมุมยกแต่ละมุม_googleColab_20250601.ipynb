## พลอตมุมของกรมอุตุนิยมวิทยา ตามข้อมูล uf ของสถานีพิษณุโลกในปี 2021

```python
# STEP 1: Install required packages
# !pip install -q wradlib h5py netCDF4

import wradlib as wrl
import matplotlib.pyplot as plt
import matplotlib as mpl
import numpy as np
import warnings

warnings.filterwarnings("ignore")

# STEP 3: Define radar parameters
nrays = 360
nbins = 500  # Changed from 480 to 500 for 250 km range
range_res = 500.0  # meters
ranges = np.arange(nbins) * range_res

# Radar site location in Thailand (TMD radar)
sitecoords = (100.217977, 16.775380, 45.0)  # (lon, lat, height in meters) # ใส่ค่าลองจิจูด, ละติจูด, ความสูงจากระดับน้ำทะเล ของสถานีที่ต้องการตรงนี้ครับ

# Beamwidth of the radar
beamwidth = 1.0

# VCP-12 standard elevation angles (in degrees)
elevs = [0.5, 1.5, 2.4, 3.4] # แก้ตามค่ามุมของเรดาร์พิษณุโลกในปี 2021

# Standard Plot (in meters)
print("Plotting radar scan strategy in meters...")
fig, ax = plt.subplots(figsize=(12, 8))
ax = wrl.vis.plot_scan_strategy(ranges, elevs, sitecoords, ax=ax)
plt.title('TMD Radar Scan Strategy - Thailand\n(Phitsanulok Elevation Pattern)', fontsize=14)
plt.tight_layout()
plt.show()

# Alternative Plot in Kilometers
print("\nPlotting radar scan strategy in kilometers...")
fig, ax = plt.subplots(figsize=(12, 8))
ax = wrl.vis.plot_scan_strategy(ranges, elevs, sitecoords, units="km", ax=ax)
plt.title('TMD Radar Scan Strategy - Thailand\n(Phitsanulok Elevation Pattern)', fontsize=14)
plt.tight_layout()
plt.show()

# Print some information about the radar setup
print(f"\nRadar Configuration:")
print(f"Location: {sitecoords[1]:.3f}°N, {sitecoords[0]:.3f}°E, {sitecoords[2]}m ASL")
print(f"Maximum range: {ranges[-1]/1000:.1f} km")
print(f"Range resolution: {range_res} m")
print(f"Number of elevation angles: {len(elevs)}")
print(f"Elevation angles: {elevs}°")
print(f"Beamwidth: {beamwidth}°")
```

## พลอตกราฟ VCP 12, VCP 215, VCP 31, VCP 35
ใช้ข้อมูลจากเวป https://www.noaa.gov/jetstream/vcp_max

```python
# STEP 1: Install required packages
# !pip install -q wradlib h5py netCDF4

import wradlib as wrl
import matplotlib.pyplot as plt
import matplotlib as mpl
import numpy as np
import warnings

warnings.filterwarnings("ignore")

# STEP 3: Define radar parameters
nrays = 360
nbins = 500  # Changed from 480 to 500 for 250 km range
range_res = 500.0  # meters
ranges = np.arange(nbins) * range_res

# Radar site location in Thailand (TMD radar)
sitecoords = (100.217977, 16.775380, 45.0)  # (lon, lat, height in meters)

# Beamwidth of the radar
beamwidth = 1.0

# Define all VCP elevation angles based on NOAA specifications https://www.noaa.gov/jetstream/vcp_max
vcp_patterns = {
    'VCP 12': {
        'elevs': [0.5, 0.9, 1.3, 1.8, 2.4, 3.1, 4.0, 5.1, 6.4, 8.0, 10.0, 12.5, 15.6, 19.5],
        'description': 'Precipitation Mode - Fast (4.5 min)\nRapid severe weather detection',
        'mode': 'Precipitation',
        'scan_time': '4.5 minutes',
        'angles': 14
    },
    'VCP 215': {
        'elevs': [0.5, 0.9, 1.3, 1.8, 2.4, 3.1, 4.0, 5.1, 6.4, 8.0, 10.0, 12.5, 15.6, 19.5, 25.0],
        'description': 'Precipitation Mode - Slower (6 min)\nBest vertical resolution, general surveillance',
        'mode': 'Precipitation',
        'scan_time': '6 minutes',
        'angles': 15
    },
    'VCP 31': {
        'elevs': [0.5, 1.5, 2.5, 3.5, 4.5],
        'description': 'Clear Air Mode - Long Pulse (10 min)\nHigh sensitivity for light precipitation',
        'mode': 'Clear Air',
        'scan_time': '10 minutes',
        'angles': 5
    },
    'VCP 35': {
        'elevs': [0.5, 0.9, 1.3, 1.8, 2.4, 3.1, 4.0, 5.1, 6.4],
        'description': 'Clear Air Mode - Default (7 min)\nQuiet weather operations',
        'mode': 'Clear Air',
        'scan_time': '7 minutes',
        'angles': 9
    }
}

# Plot each VCP pattern individually using your working code structure
for vcp_name, vcp_data in vcp_patterns.items():
    elevs = vcp_data['elevs']

    print(f"\nPlotting {vcp_name} scan strategy...")

    # Plot in kilometers (your working version)
    fig, ax = plt.subplots(figsize=(14, 8))
    ax = wrl.vis.plot_scan_strategy(ranges, elevs, sitecoords, units="km", ax=ax)

    # Create detailed title
    title = f'Radar Scan Strategy \n{vcp_name}: {vcp_data["description"]}'
    plt.title(title, fontsize=14, fontweight='bold')

    # Add information text box
    info_text = f'Mode: {vcp_data["mode"]}\n'
    info_text += f'Scan Time: {vcp_data["scan_time"]}\n'
    info_text += f'Elevation Angles: {vcp_data["angles"]}\n'
    info_text += f'Elevations: {elevs}°\n'
    info_text += f'Max Elevation: {max(elevs)}°'

    ax.text(0.02, 0.98, info_text, transform=ax.transAxes,
            verticalalignment='top', bbox=dict(boxstyle='round',
            facecolor='white', alpha=0.9), fontsize=11)

    plt.tight_layout()
    plt.show()

    # Print configuration for this VCP
    print(f"\n{vcp_name} Configuration:")
    print(f"Location: {sitecoords[1]:.3f}°N, {sitecoords[0]:.3f}°E, {sitecoords[2]}m ASL")
    print(f"Maximum range: {ranges[-1]/1000:.1f} km")
    print(f"Range resolution: {range_res} m")
    print(f"Number of elevation angles: {len(elevs)}")
    print(f"Elevation angles: {elevs}°")
    print(f"Beamwidth: {beamwidth}°")
    print(f"Mode: {vcp_data['mode']}")
    print(f"Scan Time: {vcp_data['scan_time']}")
    print("-" * 50)

# Create comparison plot (2x2 grid)
print("\nCreating VCP comparison plot...")
fig, axes = plt.subplots(2, 2, figsize=(20, 16))
axes = axes.flatten()

for i, (vcp_name, vcp_data) in enumerate(vcp_patterns.items()):
    ax = axes[i]
    elevs = vcp_data['elevs']

    # Plot using your working method
    ax = wrl.vis.plot_scan_strategy(ranges, elevs, sitecoords, units="km", ax=ax)

    # Customize each subplot
    ax.set_title(f'{vcp_name}\n{vcp_data["description"]}',
                fontsize=12, fontweight='bold', pad=15)
    ax.grid(True, alpha=0.3)

    # Add VCP information
    info_text = f'Mode: {vcp_data["mode"]}\n'
    info_text += f'Scan Time: {vcp_data["scan_time"]}\n'
    info_text += f'Elevation Angles: {vcp_data["angles"]}\n'
    info_text += f'Max Elev: {max(elevs)}°'

    ax.text(0.02, 0.98, info_text, transform=ax.transAxes,
            verticalalignment='top', bbox=dict(boxstyle='round',
            facecolor='white', alpha=0.8), fontsize=9)

plt.suptitle('WSR-88D Volume Coverage Patterns (VCP) Comparison\n Radar - scan stretegy',
             fontsize=16, fontweight='bold', y=0.95)
plt.tight_layout()
plt.show()

# Print summary comparison
print("\n" + "="*70)
print("WSR-88D Volume Coverage Patterns Summary")
print("="*70)
for vcp_name, vcp_data in vcp_patterns.items():
    print(f"\n{vcp_name}:")
    print(f"  Mode: {vcp_data['mode']}")
    print(f"  Scan Time: {vcp_data['scan_time']}")
    print(f"  Number of Elevations: {vcp_data['angles']}")
    print(f"  Elevation Angles: {vcp_data['elevs']}°")
    print(f"  Max Elevation: {max(vcp_data['elevs'])}°")
    purpose = vcp_data['description'].split('\n')[1]
    print(f"  Purpose: {purpose}")

print("\n" + "="*70)
print("Key Differences:")
print("• VCP 12: Fastest precipitation mode for severe weather (14 elevations)")
print("• VCP 215: Best vertical resolution with extra 25° elevation (15 elevations)")
print("• VCP 31: Long pulse mode for detecting light precipitation (5 elevations)")
print("• VCP 35: Default clear air mode for quiet weather (9 elevations)")
```
